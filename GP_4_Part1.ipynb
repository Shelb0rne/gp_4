#### Часть 1. A/B-тест

Для выполнения первой части группового проекта вам предоставляется датасет с результатами проведения некоторого A/B-тестирования в компании из сферы образования.


Датасет состоит из пяти столбцов:

* `user_id` — уникальный идентификатор пользователя;

* `timestamp` — время посещения пользователем страницы;

* `group` — группа эксперимента, к которой был случайно отнесён пользователь:

  *   `control` — контрольная группа,
  *   `treatment` — экспериментальная группа;

* `landing_page` — версия лендинга, которую увидел пользователь:

  *   `old_page` — старая версия страницы,
  *   `new_page` — новая версия страницы;

* `converted` — бинарный признак, показывающий, совершил ли пользователь целевое действие (целевым действием является регистрация на курс).








По замыслу A/B-теста контрольной группе (`control`) должна была показываться старая версия лендинга — `old_page`; экспериментальной группе (`treatment`) должна была показываться новая версия лендинга — `new_page`.

Вам следует:
- понять структуру данных, провести EDA и выбрать метрику, с которой будете работать;
- проверить корректность проведения тестирования (посмотреть на однородность групп; задуматься об A/A-тесте; не забыть, что в нашем распоряжении есть еще и временные метки);

- выбрать и обосновать статистический тест;
- решить, нужно ли устаналивать значения каких-либо параметров до начала анализа тестирования;
- оценить разницу в результатах и сделать выводы о практической значимости эксперимента;
- провести дополнительный анализ при необходимости (например, с помощью [bootstrap](https://habr.com/ru/companies/X5Tech/articles/679842/)  или через анализ чувствительности);

# Выбор основной метрики

У нас есть колонка converted это целевое действие на лендинге (конверсия), 1 совершил, 0 нет

- она бинарная, поэтому для сравнения долей между группами хорошо подходят тесты для пропорций, которые мы и используем



# A/A-тест

```python
aa = df_clean[df_clean['group'] == 'control'].copy()
aa = aa.sample(frac=1, random_state=42).reset_index(drop=True)   # перемешали
```

```python
half = len(aa) // 2
aa['fake_group'] = ['A'] * half + ['B'] * (len(aa) - half)
```

```python
group_a = aa[aa['fake_group'] == 'A']['converted']
group_b = aa[aa['fake_group'] == 'B']['converted']
aa_cr = aa.groupby('fake_group')['converted'].mean() * 100
aa_cr.round(3)
```

Видно, что конверсии в фейковых группах A и B почти одинаковые но проверим формально тем же z-тестом

```python
na, nb_ = len(group_a), len(group_b)
pooled_aa = (group_a.sum() + group_b.sum()) / (na + nb_)
se_aa = np.sqrt(pooled_aa * (1 - pooled_aa) * (1 / na + 1 / nb_))
```

```python
z_aa = (group_b.mean() - group_a.mean()) / se_aa
p_aa = 2 * min(stats.norm.cdf(z_aa), 1 - stats.norm.cdf(z_aa))
```

```python
round(z_aa, 3)
```

```python
round(p_aa, 4)
```

p-value получился большой, значит между двумя случайными половинками контрольной группы нет значимой разницы ура так и должно быть

# Гипотезы

так как изначально мы не знаем какое влияние окажет новая версия страницы, лучше возьмем двусторонюю гипотезу и проверим есть ли оно вообще

Таким образом наши гипотезы:

Н0: cr контрольной и экспериментальной группы не отличаются (cr_control = cr_treatment)

Н1: cr контрольной и экспериментальной группы отличаются (cr_control != cr_treatment)

#Cтат тест


В качестве стат теста мы выбрали z-test тк

1. Метрика бинарная
2. Группы независимы
3. Размеры групп большие, примерно 145 тысяч
4. При больших выборках распределение выборочной доли можно аппроксимировать нормальным распределением, что согласуется с идеей z-test


Для начала зафиксируем уровень значимости альфа = 0.05

про z-test почти вся информация/код и тд с семинара (Семинар 9-10. Statistics in Data Analysis)


image.png

```python
control = df_clean[df_clean['group'] == 'control']['converted']
treatment = df_clean[df_clean['group'] == 'treatment']['converted']
```

```python
control_sample = len(control) #кол-во в контрол
treatment_sample = len(treatment) #кол-во в тритмент
```

```python
mean_control = control.mean() #конверсия для контрол
mean_treatment = treatment.mean()  # конверсия для тритмент
```

тк в нулевой гипотезе мы предполагем, что изменений нет, то есть конверсии одинаковые, то для z теста просто найдем общую

```python
main_cr = (control.sum() + treatment.sum())/ (control_sample + treatment_sample)
main_cr
```

```python
se_ct = np.sqrt(main_cr * (1 - main_cr) *(1 /control_sample + 1 /treatment_sample))#стандартная ошибка
```

```python
diff = mean_treatment - mean_control # разность средних значений двух экспериментов
z = diff /se_ct # вычисляем з статистику
z_quantile = stats.norm.cdf(z)
print('diff=%.5f z=%.2f q=%.5f' % (diff, z, z_quantile))
```

```python
p_value = 2 *min(z_quantile, 1 - z_quantile)
p_value
```

```python
x = np.arange(-3, 3, 0.025)
plt.plot(x, stats.norm.pdf(x))
plt.axvline(z, color='k', linestyle='--')
plt.fill_between(x,stats.norm.pdf(x), where=(x <-abs(z)) | (x >abs(z)))
plt.title('z-distribution diff=%.5f z=%.2f p-value=%.5f' % (diff, z, p_value));
```

Выводы:

По результатам Z теста для двух долей значение составило -1.31, а p-value = 0.19011, тк p-value больше уровня значимости 0.05, у нас нет оснований отвергнуть нулевую гипотезу о равенстве конверсий в группах

Наблюдаемая разница конверсий составила -0.00158,однако эта разница не является статистически значимой и может быть объяснена случайными колебаниями

Следовательно, на основании имеющихся данных нельзя утверждать, что новая страница улучшает конверсию!

#Bootstrap

https://habr.com/ru/companies/X5Tech/articles/679842/$0 - отсюда брали инфу и код

Z тест показал, что между группами нет значимой разницы, но мы допом проверим это через бутстреп

В нашей выборке тритмент оказался хуже контрол, но может это случайность именно этой выборки, поэтому бутстреп поможет нам несколько раз пособирать выборки и покажет как разница между коверсиями колеблется



```python
alpha = 0.05   # уровень значимости
```

```python
B = 10000

control_values = control.values #сами элементы для выборок
treatment_values = treatment.values
diff = mean_treatment - mean_control
bootstrap_diff = []
for _ in range(B):
    bootstrap_control = np.random.choice(control_values, control_sample, replace=True) #тру, чтобы значения могли в выборки попать несколько раз
    bootstrap_treatment = np.random.choice(treatment_values, treatment_sample, replace=True)
    bootstrap_diff.append(bootstrap_treatment.mean() - bootstrap_control.mean())
std = np.std(bootstrap_diff)
print(diff)
print(std)
```

код работает не быстро, но у нас и 10000 вариантов или подвыборок...


теперь построим дов интервалы для разницы конверсий, их у нас будет 2, подробно распишу каждый

1) нормальный дов интервал, берем крит значение из нормального распределения и получается разница +- крит знач. * стандартное откл

2) открезаем по 2,5% самых больших и самых маленьких значений


```python
def get_normal_ci(bootstrap_diff, diff, alpha):
    z = stats.norm.ppf(1 - alpha / 2)
    std = np.std(bootstrap_diff)
    left, right = diff - z * std, diff + z * std
    return left, right
def get_percentile_ci(bootstrap_diff,alpha):
    left, right = np.quantile(bootstrap_diff, [alpha /2, 1 - alpha /2])
    return left, right
```

```python
normal_ci = get_normal_ci(bootstrap_diff, diff, alpha)
percentile_ci = get_percentile_ci(bootstrap_diff, alpha)
print(normal_ci)
print(percentile_ci)
```

```python
plt.figure(figsize=(16, 9))
plt.hist(bootstrap_diff, bins=50, density=True, alpha=0.6)
plt.axvline(diff, color='black',linestyle='--', label='Разница в выборке')
plt.axvline(0, color='grey',linestyle='--', label='0')

plt.axvline(normal_ci[0], color='red',linestyle='-.', label='Норм дов инт')
plt.axvline(normal_ci[1], color='red',linestyle='-.')

plt.axvline(percentile_ci[0], color='green', linestyle='-', label='Квантильный дов инт')
plt.axvline(percentile_ci[1], color='green', linestyle='-')

plt.title('Разницы в выборках бутстреп')
plt.legend()
```

вот и получились наши интервалы, они включают 0, это нам уже говорит о том, что возможная разница между группами может быть 0, то есть мы не можем уверенно сказать, что новая страница изменила конверсию, тоже самое нам и сказал z тест

# Power-анализ и MDE

```python
power_analysis = NormalIndPower()
n_per_group = min(control_sample, treatment_sample)
baseline = mean_control   # текущая конверсия для контрол
```

```python
mde_h = power_analysis.solve_power(nobs1=n_per_group, alpha=0.05, power=0.8, alternative='two-sided')
mde_h
```

```python
def h_to_p2(p2, p1, target_h):
    return 2*np.arcsin(np.sqrt(p2)) - 2*np.arcsin(np.sqrt(p1)) - target_h
```

```python
p2_mde = brentq(h_to_p2, baseline, 0.5, args=(baseline, mde_h))
mde_abs = p2_mde - baseline
```

```python
pd.Series({
    'Конверсия': round(baseline * 100, 3),
    'Конверсия с мде': round(p2_mde * 100, 3),
    'Мде': round(mde_abs * 100, 3),
    'Мде отн': round(mde_abs / baseline * 100, 2)
})
```

получается что при нашем размере выборки мы способны с вероятностью 80% обнаружить эффект, начиная с роста примерно +0.34 с 12.04% до 12.38%

а наблюдаемая разница у нас всего -0.16 пп, причём отрицательная те она сильно меньше MDE. Это значит, что выборки точно хватало если бы новая страница давала хоть сколько-нибудь заметный прирост, мы бы его увидели. Раз не увидели то эффекта действительно нет
