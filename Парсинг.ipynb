```python
import requests
from bs4 import BeautifulSoup
from time import sleep
from tqdm import tqdm
import pandas as pd
import re
import random
```

```python
url = 'https://auto.drom.ru/audi/all/'
page = requests.get(url)
page
```

```python
soup = BeautifulSoup(page.text, 'lxml') #тут смотрю структуру
print(soup.prettify())
```

```python
def get_links(make, p):
    url = f'https://auto.drom.ru/{make}/all/page{p}/'

    proxies = {
      'http': f'http://barttan_cs91l:Pamagiti1234=@unblock.oxylabs.io:60000',
      'https': f'https://barttan_cs91l:Pamagiti1234=@unblock.oxylabs.io:60000',
    }

    headers = {
        'X-Oxylabs-Session-Id': str(random.randint(0, 999999999999)),
        'x-oxylabs-geo-location': 'Russia'
    }

    page = requests.get(
        url,
        verify=False,
        proxies=proxies,
        headers=headers
    )

    soup = BeautifulSoup(page.text, 'lxml')

    links = []
    for a in soup.find_all('a'):
        href = a.get('href')
        if href and re.match(r'https?://auto\.drom\.ru/.+/.+/.+/\d+\.html', href):
            links.append(href)

    return list(set(links))
```

```python
makes_batch = ['toyota', 'hyundai', 'kia', 'cheryexeed']
links_1 = []

for make in makes_batch:
    print(f'собираю {make}')
    for p in tqdm(range(1, 41)):
        try:
            links_1.extend(get_links(make, p))
            sleep(4)
        except:
            print(f'ошибка {make} {p}')

pd.DataFrame({'url': list(set(links_1))}).to_csv('batch_1.csv', index=False)
print(f'пачка 1: {len(set(links_1))}')
```
